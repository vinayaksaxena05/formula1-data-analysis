```python
import fastf1
import pandas as pd
from pathlib import Path

fastf1.Cache.enable_cache('../cache')
```

```python
year = 2025

ROOT_DIR = Path.cwd().parent.resolve()
raw_folder = (ROOT_DIR/"data"/"raw"/f"{year}")
races = [race for race in raw_folder.iterdir() if race.is_dir]

all_telemetry = []

for round_num, race_folder in enumerate(races, start=1):
    telemetry_folder = (race_folder/"telemetry")
    drivers = [driver for driver in telemetry_folder.iterdir() if driver.is_dir]
    race_name = race_folder.name
    for driver in drivers:
        driver_name = driver.name.replace(".csv","")
        driver_telemetry = pd.read_csv(driver)
        driver_telemetry["Race"] = race_name
        driver_telemetry["Round"] = round_num
        driver_telemetry["Season"] = year 
        driver_telemetry["Driver"] = driver_name
        required_columns = ["Speed","nGear","Throttle","Brake","DRS","Distance","RPM"]
        
        driver_telemetry = driver_telemetry[driver_telemetry["Race","Round","Season","Driver",required_columns]]
        
        all_telemetry.append(driver_telemetry)
        
#===============================================================================        Incomplete For now       ======================================================================================= 
        
    
```
