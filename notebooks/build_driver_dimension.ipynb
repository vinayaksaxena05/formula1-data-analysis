```python
from pathlib import Path
import pandas as pd

ROOT_DIR = Path.cwd().parent.resolve()

year = 2025

raw_folder = ROOT_DIR / "data" / "raw" / str(year)

races = [race for race in raw_folder.iterdir() if race.is_dir()]

all_drivers = []

for race_folder in races:
    results_file = race_folder / "results.csv"

    results = pd.read_csv(results_file)

    drivers = results[
        ["DriverNumber", "Abbreviation", "FullName", "TeamName"]
    ]

    all_drivers.append(drivers)

dim_drivers = pd.concat(all_drivers, ignore_index=True)

dim_drivers = dim_drivers.drop_duplicates(subset = ["DriverNumber"])

output_folder = ROOT_DIR / "data" / "processed"
output_folder.mkdir(parents=True, exist_ok=True)

output_file = output_folder / f"dim_drivers_{year}.csv"

dim_drivers.to_csv(output_file, index=False)

    
    
    
    
```
