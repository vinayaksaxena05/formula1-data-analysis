```python
import pandas as pd
import fastf1
from pathlib import Path

fastf1.Cache.enable_cache('../cache')
```

```python
year = 2025

ROOT_DIR = Path.cwd().parent

print(f"Project Root: {ROOT_DIR}")

stints = pd.read_csv(ROOT_DIR/"data"/"processed"/"fact_driver_stints.csv")

stints = stints.sort_values(['Race','Driver','Stint'])
stints["PitLap"] = (stints.groupby(["Race","Driver"])["Laps"].transform(lambda x:x.cumsum().shift()))
pitstops = stints[stints["Stint"]>1]
pitstops = pitstops[["Race","Driver","Stint","PitLap","Compound"]]

pitstops.to_csv(ROOT_DIR/"data"/"processed"/f"pitstop_summary_{year}.csv", index=False)
```
