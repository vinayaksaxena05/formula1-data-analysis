```python
from pathlib import Path
import pandas as pd
import fastf1

ROOT_DIR = Path.cwd().parent.resolve()

year = 2025

processed_dir = ROOT_DIR / "data" / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)

schedule = fastf1.get_event_schedule(year)

dim_races = schedule.loc[
    schedule["RoundNumber"] > 0,
    [
        "RoundNumber",
        "Country",
        "Location",
        "EventName",
        "OfficialEventName",
        "EventDate",
        "EventFormat"
    ]
].copy()

dim_races.insert(0, "RaceID", range(1, len(dim_races) + 1))

dim_races["SprintWeekend"] = dim_races["EventFormat"].apply(
    lambda x: "Y" if "sprint" in str(x).lower() else "N"
)

dim_races = dim_races.drop(columns=["EventFormat"])

output_file = processed_dir / f"dim_races_{year}.csv"

dim_races.to_csv(output_file, index=False)

dim_races.head()
```
