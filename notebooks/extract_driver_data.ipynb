```python
from pathlib import Path
import fastf1

schedule = fastf1.get_event_schedule(2025)
races = schedule[schedule['EventName'] != 'Pre-Season Testing']

for _, row in races.iterrows():

    race_name = row['EventName'].replace(" Grand Prix", "")
    folder_name = race_name.lower().replace(" ", "_")

    print(f"\nProcessing {race_name}")

    telemetry_folder = Path(
        f"../data/raw/2025/{folder_name}/telemetry"
    )
    telemetry_folder.mkdir(parents=True, exist_ok=True)

    session = fastf1.get_session(
        2025,
        row['RoundNumber'],
        'R'
    )

    print("Loading session data...")
    session.load()

    print(f"Found {len(session.drivers)} drivers")

    for driver in session.drivers:

        driver_name = session.get_driver(driver)['Abbreviation']

        file_path = telemetry_folder / f"{driver_name}.csv"

        if file_path.exists():
            print(f"Skipping {driver_name}")
            continue

        driver_laps = session.laps.pick_drivers(driver)

        if driver_laps.empty:
            print(f"No lap data for {driver_name}")
            continue

        try:
            print(f"Saving telemetry for {driver_name}")

            telemetry = driver_laps.get_telemetry()

            telemetry.to_csv(
                file_path,
                index=False
            )

        except ValueError as e:
            print(
                f"Could not get telemetry for "
                f"{driver_name}: {e}"
            )
            continue

    print(f"Finished {race_name}")

print("\nAll races processed")
```
