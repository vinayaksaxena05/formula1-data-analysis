```python
## Packages
import fastf1
import matplotlib.pyplot as plt
import pandas as pd
import pathlib as Path
fastf1.Cache.enable_cache("../cache")
```

```python
schedule = fastf1.get_event_schedule(2025)

races = schedule[schedule['EventName']!='Pre-Season Testing']

for _,row in races.iterrows():
    
    race_name = row['EventName'].replace(" Grand Prix", "")
    folder_name = race_name.lower().replace(" ","_")
    print(f"Processing {race_name}...")
    session = fastf1.get_session(2025,row['RoundNumber'],'R')
    session.load()

    race_folder = Path(f"../data/raw/2025/{folder_name}")
    if (race_folder / "laps.csv").exists():
        print(f"Skipping {race_name}")
        continue
    race_folder.mkdir(parents=True, exist_ok=True)

    session.laps.to_csv(race_folder/"laps.csv",index=False)
    session.results.to_csv(race_folder/"results.csv",index=False)
    session.weather_data.to_csv(race_folder/"weather.csv",index=False)

    
```
